```python
def canonicalize_categories(category_list: list[str], canonical_taxonomy: str,
                              batch_size: int = 300) -> dict:
    """Maps a long list of raw event_type labels to a fixed canonical taxonomy,
    in batches, to avoid single-call output truncation."""
    full_mapping = {}

    for i in range(0, len(category_list), batch_size):
        batch = category_list[i:i + batch_size]
        prompt = f"""Here is a fixed canonical taxonomy of event categories:

{canonical_taxonomy}

Map each of the following ORIGINAL labels to the single best-fitting
canonical category above. Return ONLY a JSON object (no markdown fences,
no explanation) mapping each original label to its canonical category.

Labels to map:
{batch}
"""
        response = client.messages.create(
            model="claude-sonnet-4-6",
            max_tokens=4096,
            messages=[{"role": "user", "content": prompt}]
        )
        batch_mapping = json.loads(response.content[0].text)
        full_mapping.update(batch_mapping)
        print(f"Mapped batch {i // batch_size + 1}: {len(batch)} labels")

    return full_mapping
```

```python
import json
import anthropic
from dotenv import load_dotenv
load_dotenv()
client = anthropic.Anthropic()

canonical_taxonomy = """
1. capacity_expansion - physical capacity, capex, manufacturing scale-up
2. strategic_partnership - alliances, collaborations, joint ventures
3. financial_performance - earnings, revenue results, beats/misses
4. revenue_growth - revenue growth, records, acceleration
5. product_launch_and_development - new products, roadmaps, innovations
6. order_dynamics - wins, losses, cancellations, backlogs
7. market_dynamics - demand surges, declines, recoveries, share shifts
8. pricing_and_margins - price changes, margin expansion/compression
9. supply_chain - disruptions, partnerships, diversification, constraints
10. ai_demand - AI-specific demand, infrastructure, servers
11. competitive_dynamics - threats, advantages, positioning
12. analyst_and_market_sentiment - upgrades, downgrades, targets, stock moves
13. geopolitical_and_regulatory - tariffs, trade, export controls, policy
14. corporate_events - M&A, IPO, restructuring, leadership, dividends
15. technology_advancement - tech breakthroughs, process nodes, R&D
"""

category_list = signals_df['event_type'].value_counts().index.tolist()
category_mapping = canonicalize_categories(category_list, canonical_taxonomy)
```

```python
signals_df['event_type_canonical'] = signals_df['event_type'].map(category_mapping)
signals_df['event_type_canonical'].isna().sum()
```
